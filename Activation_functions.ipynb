# Summary of commonly used activation functions in Machine Learning

## Sigmoid Function (logistic function)
#### Sigmoid is commonly used in binary classification problems, where the goal is to classify input into one of two classes

```python
import numpy as np
```

```python
def compute_sigmoid(x):
    phi_of_x = 1. / (1. + np.exp(-np.clip(x, -250, 250)))
    return phi_of_x
```

```python
def derivate_sigmoid(phi_of_x):
    return phi_of_x * (1. - phi_of_x)
```

## ReLU Function (Rectified Linear Unit)
#### ReLU is the activation function very popular in neural networks

```python
def compute_relu(x):
    phi_of_x = np.where(x <= 0, 0, x)
    return phi_of_x
```

```python
def derivate_relu(phi_of_x):
    return np.where(phi_of_x <= 0, 0, 1)
```

## Leaky ReLU
link source: https://arxiv.org/pdf/1505.00853

```python
def compute_leaky_relu(x, alpha=0.01): #alpha is fixed
    return np.where(x >= 0, x, x * alpha)
```

```python
def derivative_leaky_relu(x, alpha=0.01): #alpha is fixed
    return np.where(x >= 0, 1, alpha)
```

## Bent Indentity
link source: https://www.gabormelli.com/RKB/Bent_Identity_Activation_Function

```python
def compute_bent_identity(x):
    return ((np.sqrt(x ** 2 + 1) - 1) / 2) + x
```

```python
def derivative_bent_identity(x):
    return (x / (2 * np.sqrt(x ** 2 + 1))) + 1
```

## Hard Tanh

```python
def compute_hard_hyperbolic_tangent(x):
    return np.clip(x, -1, 1)
```

```python
def derivative_hard_hyperbolic_tangent(x):
    return np.where((x <= 1) & (x >= -1), 1, 0)
```
